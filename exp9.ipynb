```python
import pandas as pd
```

```python
df = pd.read_csv('titanic.csv',usecols=['Age', 'Fare', 'Survived', 'Cabin'])
df
```

```python
df.isnull().sum()
```

```python
df['Survived'].fillna(1, inplace=True)
```

```python
df.isnull().sum
```

```python
df['Fare'].fillna(method="ffill",inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Cabin']=df['Cabin'].bfill()
```

```python
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
print(df.isnull().sum())
```
