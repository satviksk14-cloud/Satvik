```python
import pandas as pd
df=pd.read_csv('CreditCard_DV - Copy.csv')
```

```python
import matplotlib.pyplot as plt
fig=plt.figure(figsize=[5,6])
ax=fig.add_axes([0,0,1,1])
ax.scatter(df['Credit_Limit'],df['Avg_Purchase'])
ax.set_title("Credit limit vs Avg purchase",fontsize=16)
ax.set_xlabel("Credit limit",fontsize=12)
ax.set_ylabel("Avg purchase",fontsize=12)
fig.show()
```

```python
import matplotlib.pyplot as plt
df_list=[df['Credit_Limit'],df['Avg_Purchase']]
fig=plt.figure(figsize=[6,5])
ax=fig.add_axes([0,0,1,1])
ax.set_title(['Credit_Limit and Avg_Purchase'])
ax.boxplot(df_list)
ax.set_xticklabels(['Credit_Limit','Avg_Purchase'])
plt.show()
```

```python
grouped_df=df.groupby('Income_Category').size().reset_index(name='count')
fig=plt.figure(figsize=[7,5])
ax=fig.add_axes([0,0,1,1])
ax.set_title("No.of clusters by income category",fontsize=16)
ax.set_xlabel('Income category',fontsize=12)
ax.set_ylabel('count',fontsize=12)
ax.bar(grouped_df['Income_Category'],grouped_df['count'])
plt.tight_layout()
plt.show()
```

```python
fig=plt.figure(figsize=[8,6])
ax=fig.add_axes([0,0,1,1])
ax.hist(df['Total_Trans_Amt'],bins=20,edgecolor='black')
ax.set_title('Distribution of total transaction amount',fontsize=17)
ax.set_xlabel('Total transaction amount',fontsize=12)
ax.set_ylabel('Frequency',fontsize=12)
plt.show()
```

```python
pie_df=pd.DataFrame()
pie_df['count']=df['Attrition_Flag'].value_counts()
pie_df=pie_df.reset_index()
pie_df.rename(columns={'index':'customers'},inplace=True)
pie_df
fig=plt.figure(figsize=[6,4])
ax=fig.add_axes([0,0,1,1])
ax.set_title('Attrition_Flag')
ax.pie(pie_df['count'],labels=pie_df['Attrition_Flag'],autopct='%0.1f%%')
fig.show()
```
